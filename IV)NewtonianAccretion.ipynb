The flux equation for the accretion disk is given by:

$$
F(r) = \frac{3}{8\pi} \frac{GM \dot{M}}{r^3} \left( 1 - \sqrt{\frac{r_{\text{in}}}{r}} \right)
$$

We derive this equation step by step.

### 1. Mass Conservation in the Disk

The mass accretion rate $(\dot{M})$ is constant throughout the disk, and the mass flux is given by:

$$
\dot{M} = 2\pi r \Sigma v_r
$$

where $r$ is the radial distance, $\Sigma$ is the surface density of the disk, and $v_r$ is the radial inflow velocity of matter.

### 2. Energy Dissipation in the Disk

The energy dissipated per unit area of the disk, D(r), is proportional to the viscous torque, which is related to the angular momentum flux. The specific angular momentum at radius \(r\) is:

$$
j(r) = \sqrt{GM r}
$$

The torque at radius \(r\) is:

$$
G(r) = \dot{M} \frac{j(r)}{2\pi} = \dot{M} \frac{\sqrt{GM r}}{2\pi}
$$

The rate of energy dissipation D(r) is the derivative of the torque with respect to \(r\):

$$
D(r) = \frac{1}{r} \frac{dG(r)}{dr}
$$

Taking the derivative of G(r), we get:

$$
D(r) = \frac{\dot{M}}{4\pi} \frac{GM}{r^3}
$$

### 3. Boundary Condition for Inner Disk Edge

At the inner edge of the disk, $r = r_{in}$, the material falls into the black hole, effectively removing all angular momentum. Thus, we modify the dissipation at the inner edge by adding a correction term:

$$
F(r) = D(r) \left( 1 - \sqrt{\frac{r_{\text{in}}}{r}} \right)
$$

### 4. Final Expression

Substituting the expression for \(D(r)\) into the equation for \(F(r)\), we get:

$$
F(r) = \frac{3}{8\pi} \frac{GM \dot{M}}{r^3} \left( 1 - \sqrt{\frac{r_{\text{in}}}{r}} \right)
$$

```python
import numpy as np
from astropy.constants import G
```

```python
M = 2 # Mass (arbitrary units)
Mdot = 10 # Mass accretion rate (arbitrary units)
r_in = 3 # Interior radius
```

```python
r = 4 # Radius at which to evaluate the equation
print(3. * G.value * M * Mdot / (8. * np.pi * r**3) * (1. - np.sqrt(r_in / r)))
```

```python
# Define a function called F with argument r
def F(r):
    flux = 3. * G.value * M * Mdot / (8. * np.pi * r**3) * (1. - np.sqrt(r_in / r))    # Calculate the flux
    return flux      # Return the flux as an output
```

```python
radii = np.arange(1, 50, 0.5)
print(radii)
```

```python
F(radii)
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

# Plot the data!
plt.plot(radii, F(radii))
plt.xlabel("Radius") # Set the label on the x axis
plt.ylabel("Flux") # Set the label on the y axis
plt.yscale("log")# Use a log-scale on the y axis instead of linear
plt.show()
```

Now we want to explore the chages to the plot on varying the mass $M$ , $ \dot{M}$ and $r_{in}$

```python
# Define a function called F with arguments r, M, Mdot, and r_in
def F(r, M, Mdot, r_in):
    flux = 3. * G.value * M * Mdot / (8. * np.pi * r**3) * (1. - np.sqrt(r_in / r))    # Calculate the flux
    return flux      # Return the flux as an output
```

```python
radii = np.linspace(0.001, 100, 200) # Create an array from 0 to 100 with 200 data points
```

```python
data_1 = F(radii, M=10, Mdot=4, r_in=3)
data_2 = F(radii, M=20, Mdot=4, r_in=3)
data_3 = F(radii, M=30, Mdot=4, r_in=3)
# Plot the data we just made on the same graph using different labels
plt.plot(radii, data_1, label='M=10')
plt.plot(radii, data_2, label='M=20')
plt.plot(radii, data_3, label='M=30')
plt.xlabel("Radius") # Set the label on the x axis
plt.ylabel("Flux") # Set the label on the y axis
plt.yscale("log")# Use a log-scale on the y axis instead of linear
plt.legend() # Show the legend
plt.show()
```
